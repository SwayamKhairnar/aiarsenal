**Valid Sudoku**

```python
s = np.array([
    [5, 3, 4, 6, 7, 8, 9, 1, 2],
    [6, 7, 2, 1, 9, 5, 3, 4, 8],
    [1, 9, 8, 3, 4, 2, 5, 6, 7],
    
    [8, 5, 9, 7, 6, 1, 4, 2, 3],
    [4, 2, 6, 8, 5, 3, 7, 9, 1],
    [7, 1, 3, 9, 2, 4, 8, 5, 6],
    
    [9, 6, 1, 5, 3, 7, 2, 8, 4],
    [2, 8,4, 4, 1, 9, 6, 3, 5],
    [3, 4, 5, 2, 8, 6, 1, 7, 9]
])
```

we have to check weather the soduku is valid or not
so we check the row and col sum weather for each row anc col it is 45 or not if yes then valid else not 

**General Questions**

```python
b=np.sum(s,axis=1)
for i in b:
    if(i!=45):
        print("not valid")
        break
else:
    print("valid")
```

```python
c=np.sum(s,axis=0)
for i in c:
    if(i!=45):
        print("not valid")
        break
else:
    print("valid")
```

```python
#Find the highest Science mark.
np.max(data[:,2:3])
```
